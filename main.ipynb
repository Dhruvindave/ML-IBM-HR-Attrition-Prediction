#### 2.2.3 Numerical Suspicious Value Investigation

````python
"""
    To look:
    MonthlyIncome
    NumCompaniesWorked
    TotalWorkingYears
    TrainingTimesLastYear
    YearsAtCompany
    YearsInCurrentRole
    YearsSinceLastPromotion
    YearsWithCurrManager
"""
suspicious_numerical_col = ['MonthlyIncome', 'NumCompaniesWorked', 'TotalWorkingYears', 'TrainingTimesLastYear', 'YearsAtCompany', 'YearsInCurrentRole', 'YearsSinceLastPromotion', 'YearsWithCurrManager']
````

##### 2.2.3.1 Overall Analysis

For every numerical variable investigate:

* Negative values
* Zero values
* Extremely small values
* Extremely large values
* Impossible values
* Potential outliers

````python
def suspicious_values_analysis(df, numerical_features):
    results = {}

    for col in numerical_features:
        negatives = (df[col] < 0).sum()
        zeros = (df[col] == 0).sum()

        max_val = df[col].max()
        min_val = df[col].min()

        # IQR outlier detection
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)

        IQR = Q3-Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]

        results[col] = {
            "negatives": negatives,
            "zeros": zeros,
            "min": min_val,
            "max": max_val,
            "outliers": outliers
        }

    return pd.DataFrame(results).T
````

````python
suspicious_values_analysis(df, numerical_features=suspicious_numerical_col)
````

# 3. Feature Engineering

## 3.1 Feature Building

````python
def engineer_features(data: pd.DataFrame) -> pd.DataFrame:
    """
    Create meaningful modeling features from the IBM HR Attrition dataset.

    The input DataFrame is never modified.
    """
    features = data.copy()

    # ------------------------------------------------------------------
    # 1. Remove identifiers, constants, and EDA-only derived columns
    # ------------------------------------------------------------------
    drop_columns = [
        "AttritionNumeric",
        "IncomeBucket",
        "DistanceGroup",
        "TenureGroup",
        "EmployeeNumber",
        "EmployeeCount",
        "StandardHours",
        "Over18",
    ]

    features = features.drop(
        columns=drop_columns,
        errors="ignore",
    )

    # ------------------------------------------------------------------
    # 2. Career / tenure features
    # ------------------------------------------------------------------

    # Total career experience accumulated before joining the current company.
    features["YearsBeforeCurrentCompany"] = (
        features["TotalWorkingYears"] - features["YearsAtCompany"]
    )

    # Proportion of company tenure spent in the current role.
    company_tenure = features["YearsAtCompany"].replace(0, np.nan)

    features["RoleTenureRatio"] = features["YearsInCurrentRole"] / company_tenure

    # Proportion of company tenure spent under the current manager.
    features["ManagerTenureRatio"] = features["YearsWithCurrManager"] / company_tenure

    # Relative time since the last promotion.
    features["PromotionGapRatio"] = features["YearsSinceLastPromotion"] / company_tenure

    # ------------------------------------------------------------------
    # 3. Age representation
    # ------------------------------------------------------------------
    features["AgeGroup"] = pd.cut(
        features["Age"],
        bins=[17, 29, 39, 49, 60],
        labels=["<30", "30-39", "40-49", "50+"],
        include_lowest=True,
    )

    # ------------------------------------------------------------------
    # 4. EDA-driven interaction features
    # ------------------------------------------------------------------
    overtime = features["OverTime"].eq("Yes")
    young = features["Age"].lt(30)
    single = features["MaritalStatus"].eq("Single")

    features["YoungWithOverTime"] = (young & overtime).astype("int8")
    features["SingleWithOverTime"] = (single & overtime).astype("int8")
    features["YoungSingleWithOverTime"] = (young & single & overtime).astype("int8")

    return features
````

````python
# =============================================================
# Section 3: Feature Engineering
# =============================================================
engineered_df = engineer_features(df)
````

## 3.2 Feature Encoding

````python
# =============================================================
# Section 4: Target Preparation
# =============================================================
y = engineered_df["Attrition"].map({"No": 0, "Yes": 1})

if y.isna().any():
    raise ValueError("Attrition contains values other than 'No' and 'Yes'.")

X = engineered_df.drop(columns="Attrition")

# =============================================================
# Section 5: Feature-Group Identification
# =============================================================

# Ordinal features: passed through as-is (integer survey scales / levels).
ordinal_features = [col for col in ordinal_cols if col in X.columns]

# Categorical (nominal) features: start from EDA-defined list, then
# automatically pick up any remaining object / category columns created
# during feature engineering (e.g. AgeGroup).
categorical_features = [col for col in categorical_cols if col in X.columns]
categorical_features += [
    col
    for col in X.select_dtypes(include=["object", "str", "category"]).columns
    if col not in categorical_features
]
categorical_features = list(dict.fromkeys(categorical_features))  # preserve order, deduplicate

# Numeric features: all numeric columns that are not already in ordinal_features.
numeric_features = [
    col
    for col in X.select_dtypes(include=np.number).columns
    if col not in ordinal_features
]

# Validation: every column must belong to exactly one group.
unhandled_features = set(X.columns) - set(
    ordinal_features + numeric_features + categorical_features
)
if unhandled_features:
    raise ValueError(
        f"Features need an encoding strategy: {sorted(unhandled_features)}"
    )

print("Ordinal features  :", ordinal_features)
print("Numeric features  :", numeric_features)
print("Categorical features:", categorical_features)
````

````python
# =============================================================
# Section 6: Train / Test Split  (split BEFORE fitting the preprocessor)
# =============================================================
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y,
)

print(f"\nTraining set : {X_train.shape[0]} rows")
print(f"Test set     : {X_test.shape[0]} rows")
````

## 5. Preprocessing

````python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
````

````python
# =============================================================
# Section 7: Preprocessor  (ColumnTransformer)
# =============================================================
preprocessor_logistic_regression = ColumnTransformer(
    transformers=[
        # Ordinal features: integer scale values are already meaningful;
        # scale for lbfgs convergence but do not encode.
        (
            "ordinal",
            Pipeline([
                ("scaler", StandardScaler()),   # required for lbfgs convergence
            ]),
            ordinal_features,
        ),

        # Numeric features: median imputation guards against any NaN values
        # that arise from ratio calculations (e.g. division by 0 tenure).
        (
            "numeric",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler()),   # required for lbfgs convergence
            ]),
            numeric_features,
        ),

        # Categorical features: impute rare missings, then one-hot encode.
        (
            "categorical",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                (
                    "encoder",
                    OneHotEncoder(
                        handle_unknown="ignore",
                        sparse_output=False,   # sklearn >= 1.2
                    ),
                ),
            ]),
            categorical_features,
        ),
    ],
    remainder="drop",
)


preprocessor_tree_based = ColumnTransformer(
    transformers=[
        (
            "numeric",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
            ]),
            numeric_features,
        ),
        (
            "categorical",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                (
                    "encoder",
                    OneHotEncoder(
                        handle_unknown="ignore",
                        sparse_output=False,   # sklearn >= 1.2
                    ),
                ),
            ]),
            categorical_features,
        ),
    ],
    remainder="drop"
)
````

# 8. Model Evaluation

Evaluate the tuned Logistic Regression once on the held-out test set. This gives a final estimate on data that was not used to choose hyperparameters. ROC-AUC uses predicted probabilities; the other classification metrics use the pipeline's default `predict()` decision.

````python
final_predictions = tuned_logistic_regression.predict(X_test)
final_probabilities = tuned_logistic_regression.predict_proba(X_test)[:, 1]

final_metrics = pd.DataFrame(
    {
        "Score": [
            accuracy_score(y_test, final_predictions),
            precision_score(y_test, final_predictions, zero_division=0),
            recall_score(y_test, final_predictions, zero_division=0),
            f1_score(y_test, final_predictions, zero_division=0),
            roc_auc_score(y_test, final_probabilities),
        ]
    },
    index=["Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"],
)
display(final_metrics.round(4))

final_confusion = confusion_matrix(y_test, final_predictions, labels=[0, 1])
display(
    pd.DataFrame(
        final_confusion,
        index=["Actual: No Attrition", "Actual: Attrition"],
        columns=["Predicted: No Attrition", "Predicted: Attrition"],
    )
)

final_report = classification_report(
    y_test,
    final_predictions,
    labels=[0, 1],
    target_names=["No Attrition", "Attrition"],
    output_dict=True,
    zero_division=0,
)
display(pd.DataFrame(final_report).T.round(4))

tn, fp, fn, tp = final_confusion.ravel()
print(
    f"Attrition class: detected {tp} of {tp + fn} actual attrition cases; "
    f"missed {fn}. Of {tp + fp} predicted attrition cases, {tp} were correct."
)
````

# 9. Model Interpretation

Inspect the Logistic Regression coefficients after the fitted preprocessor. `get_feature_names_out()` supplies the transformed names, including individual one-hot categories, so coefficient rows remain identifiable. Positive coefficients are associated with higher predicted odds of Attrition; negative coefficients with lower odds, holding the other transformed features constant. These are model associations, not causal effects.

````python
final_feature_names = tuned_logistic_regression.named_steps[
    "preprocessor"
].get_feature_names_out()
final_coefficients = tuned_logistic_regression.named_steps["model"].coef_.ravel()

coefficient_table = pd.DataFrame(
    {
        "Feature": final_feature_names,
        "Coefficient": final_coefficients,
    }
)

top_positive_coefficients = coefficient_table.nlargest(10, "Coefficient")
top_negative_coefficients = coefficient_table.nsmallest(10, "Coefficient")

display(top_positive_coefficients.round(4))
display(top_negative_coefficients.round(4))
````

The displayed tables rank the strongest positive and negative coefficient values. Numeric and ordinal inputs are scaled by the fitted preprocessor, so their coefficients correspond to a one-standard-deviation change; one-hot rows represent the encoded category indicator. Coefficients are changes in log-odds, not direct percentage-point changes in probability. These patterns describe associations learned from this dataset and do not establish that any feature causes attrition.
